# Load Training Data

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 

df = pd.read_csv("train.csv")
df.head()
```

# EDA (Exploratory Data Analysis)

```python
# check feature types and missing values before modification
df.info()
```

```python
# check for missing values and the shape of the dataset

print(df.isnull().sum())
print(df.shape)
```

```python
# show the summary statistics of the dataset
df.describe()
```

```python
# Fill missing values with median 
# check before and after filling missing values

print("Missing values before filling:")
print(df.isnull().sum())
df['f04'].fillna(df['f04'].median(), inplace=True)
df['f06'].fillna(df['f06'].median(), inplace=True)
print("Missing values after filling:")
print(df.isnull().sum())
```

```python
# check the data types and missing values after filling missing values
df.info()
```

```python
# check for duplicate rows in the dataset
print(df.duplicated().sum())
```

```python
# drop the 'id' column as it is not useful for classification
df.drop('id',axis=1,inplace=True)
```

```python
# check the distribution of the features in the dataset

df.hist(figsize=(15, 10))
plt.tight_layout()
plt.show()
```

```python
# check the correlation and collinearity between features in the dataset

corr = df.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

# Spliting

```python
from sklearn.model_selection import train_test_split

X = df.drop(columns=['target'])
y = df['target']

X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
```

# Scaling

```python
from sklearn.preprocessing import RobustScaler
from sklearn.metrics import roc_auc_score

scaler = RobustScaler()

X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
```

# LogisticRegression --> Base Model 

```python
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter=250, random_state=42)

log_reg.fit(X_train, y_train)

y_val_pred = log_reg.predict_proba(X_val)[:, 1]

auc = roc_auc_score(y_val, y_val_pred)
print("Validation ROC-AUC:", auc)
```

# Perform cross-validation with StratifiedKFold


```python
# Perform cross-validation with StratifiedKFold

from sklearn.model_selection import StratifiedKFold
from lightgbm import LGBMClassifier

kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

auc_scores = []

for train_index, val_index in kf.split(X, y):
    X_train, X_val = X.iloc[train_index], X.iloc[val_index]
    y_train, y_val = y.iloc[train_index], y.iloc[val_index]

    scaler = RobustScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_val_scaled = scaler.transform(X_val)

    lgb_clf = LGBMClassifier(
        n_estimators=500,
        learning_rate=0.05,
        num_leaves=31,
        random_state=42
    )

    lgb_clf.fit(X_train_scaled, y_train)
    y_val_pred = lgb_clf.predict_proba(X_val_scaled)[:, 1]

    auc = roc_auc_score(y_val, y_val_pred)
    auc_scores.append(auc)

print("AUC scores per fold:", auc_scores)
print("Mean AUC:", np.mean(auc_scores))
```

# GridSearchCV for xgboost

```python
from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV

xgb_clf = XGBClassifier(
    use_label_encoder=False,
    eval_metric='auc',
    random_state=42
)

param_grid = {
    'n_estimators': [200, 500, 800],
    'max_depth': [4, 6, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

grid_search = GridSearchCV(
    estimator=xgb_clf,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=3,              
    verbose=2,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best ROC-AUC:", grid_search.best_score_)
```

# GridSearchCV for lightgbm

```python
from lightgbm import LGBMClassifier
from sklearn.model_selection import GridSearchCV

lgb_clf = LGBMClassifier(random_state=42)

param_grid = {
    'n_estimators': [200, 500, 800],
    'num_leaves': [31, 50, 70],
    'learning_rate': [0.01, 0.05, 0.1],
    'max_depth': [-1, 6, 10],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

grid_search = GridSearchCV(
    estimator=lgb_clf,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=3,              
    verbose=2,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best ROC-AUC:", grid_search.best_score_)
```

# GridSearchCV for catboost

```python
from catboost import CatBoostClassifier
from sklearn.model_selection import GridSearchCV

cat_clf = CatBoostClassifier(
    random_state=42,
    verbose=0
)

param_grid = {
    'iterations': [200, 500],
    'learning_rate': [0.01, 0.05, 0.1],
    'depth': [4, 6, 8],
    'l2_leaf_reg': [1, 3, 5],
    'subsample': [0.8, 1.0]
}

grid_search = GridSearchCV(
    estimator=cat_clf,
    param_grid=param_grid,
    scoring='roc_auc',
    cv=3,              
    verbose=2,
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

print("Best parameters (CatBoost):", grid_search.best_params_)
print("Best ROC-AUC (CatBoost):", grid_search.best_score_)
```

# RandomForest Model

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=700,       
    max_depth=15,
    min_samples_split=15,
    max_features="sqrt" , 
    min_samples_leaf=10,
    oob_score=True,
    random_state=42,
    n_jobs=-1               
)
rf.fit(X_train, y_train)

y_val_pred = rf.predict_proba(X_val)[:, 1]

auc_rf = roc_auc_score(y_val, y_val_pred)
print("Validation ROC-AUC (Random Forest):", auc_rf)
```

```python
# Feature importance from Random Forest

important = pd.Series(rf.feature_importances_ , index=X.columns)
print(important.sort_values(ascending=False))
```
